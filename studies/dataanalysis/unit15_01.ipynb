```python
url = "https://en.wikipedia.org/wiki/All-time_Olympic_Games_medal_table"
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

response = requests.get(url, headers=headers)

df_list = pd.read_html(StringIO(response.text))

summer = df_list[1].iloc[:,:6]
summer.columns = ['Nation', 'NumberOfParticipation', 'SummerGold', 'SummerSilver', 'SummerBronze', 'SummerTotal']
summer = summer.set_index('Nation')
summer.sort_values("SummerGold", ascending=False)
```

```python
summer.to_excel(r"./data/summer_olympic_medal_table.xlsx")
```
